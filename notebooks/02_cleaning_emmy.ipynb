```python
import pandas as pd # jobba med tabeller/dataframes
import numpy as np # bland annat för NaN
from pathlib import Path # hitta rätt filväg till datasetet
```

```python
# Gör att filvägen fungerar oavsett var notebooken körs ifrån
project_folder = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()

# Läser in den smutsiga filen
df = pd.read_csv(project_folder / "data" / "dirty" / "yrbs_dirty_9vars.csv")

# skapar en mindre dataframa med mina 3 variabler
emmy_cols = ["row_id", "weight_orig", "mental_health_not_good", "grade"]
df_emmy = df[emmy_cols].copy()

# Shape - visar antal rader och kolumner
print(df_emmy.shape)
```

```python
# Upptäcka konstiga värden
display(df_emmy.head(20)) # head() visar början av datan
display(df_emmy.sample(30, random_state=42)) # sample() visar slumpmässiga rader, så vi inte bara ser toppen
```

```python
# Visar antal värden, datatyper och hur många som inte är null
df_emmy.info()
```

```python
# Visar hur många riktiga saknade värden som redan finns
print(df_emmy.isnull().sum())
```

```python
# Visar om det finns identiska rader
print(df_emmy.duplicated().sum())
```

```python
# Extra viktigt - visar 'N', 'N N', tomma strängar, bokstäver i numeriska kolumner och ovanliga/extrema värden
print(df_emmy['grade'].value_counts(dropna=False))
print(df_emmy['mental_health_not_good'].value_counts(dropna=False))
print(df_emmy['weight_orig'].value_counts(dropna=False).head(30))
```

```python
# Upptäcker orimliga värden
# Försöker göra om 'weight_orig' till siffror, det som inte går blir NaN, sedan får vi statistik som mean, min, max
print(pd.to_numeric(df_emmy['weight_orig'], errors='coerce').describe())
```

```python
# Visar rader där 'grade' inte grå att läsa som siffra
display(df_emmy[
    pd.to_numeric(df_emmy['grade'], errors='coerce').isna()
].head(30))
```

```python
# Visar rader där 'mental_health...' inte grå att läsa som siffra
display(df_emmy[
    pd.to_numeric(df_emmy['mental_health_not_good'], errors='coerce').isna()
].head(30))
```

```python
# Visar rader där 'weight_orig' inte grå att läsa som siffra
display(df_emmy[
    pd.to_numeric(df_emmy['weight_orig'], errors='coerce').isna()
].head(30))
```

```python
# Kollar om vikten verkar orimlig. Gränserna 400-50 hjälper oss hitta misstänkta värden
weight_num = pd.to_numeric(df_emmy['weight_orig'], errors='coerce')
display(df_emmy[(weight_num > 400) | (weight_num < 50)].head(30))
```

```python
# df_clean = fixad data
df_clean = df_emmy.copy()
```

```python
df_clean = df_clean.replace(['', ' ', 'N', 'N N'], np.nan)
```

```python
df_clean['weight_orig'] = pd.to_numeric(df_clean['weight_orig'], errors='coerce')
df_clean['mental_health_not_good'] = pd.to_numeric(df_clean['mental_health_not_good'], errors='coerce')
df_clean['grade'] = pd.to_numeric(df_clean['grade'], errors='coerce')
```

```python
df_clean.loc[~df_clean['weight_orig'].between(50, 400), 'weight_orig'] = np.nan
```

```python
df_clean.info()
df_clean.isnull().sum()
```

```python
print('Före rensning:')
print(df_emmy.info())

print('\nEfter rensning:')
print(df_clean.info())
```
