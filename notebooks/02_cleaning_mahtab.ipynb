```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
project_folder = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
```

```python
df = pd.read_csv(project_folder / "data" / "dirty" / "yrbs_dirty_9vars.csv")
```

```python
mahtab_cols = ["row_id", "bmipct", "sex", "sleep_hours"]
df_mahtab = df[mahtab_cols].copy()

print("Shape:")
print(df_mahtab.shape)

print("\nFörsta 20 raderna:")
display(df_mahtab.head(20))

print("\n30 slumpmässiga rader:")
display(df_mahtab.sample(30, random_state=42))

print("\nInfo:")
df_mahtab.info()

print("\nMissing values:")
print(df_mahtab.isnull().sum())

print("\nDubbletter:")
print(df_mahtab.duplicated().sum())

print("\nVärden i sex:")
print(df_mahtab["sex"].value_counts(dropna=False))

print("\nVärden i sleep_hours:")
print(df_mahtab["sleep_hours"].value_counts(dropna=False))

print("\nVärden i bmipct:")
print(df_mahtab["bmipct"].value_counts(dropna=False).head(30))

print("\nBeskrivning av bmipct:")
print(pd.to_numeric(df_mahtab["bmipct"], errors="coerce").describe())

print("\nRader där sex inte är numerisk eller saknas:")
display(df_mahtab[
    pd.to_numeric(df_mahtab["sex"], errors="coerce").isna()
].head(30))

print("\nRader där sleep_hours inte är numerisk eller saknas:")
display(df_mahtab[
    pd.to_numeric(df_mahtab["sleep_hours"], errors="coerce").isna()
].head(30))

print("\nRader där bmipct inte är numerisk eller saknas:")
display(df_mahtab[
    pd.to_numeric(df_mahtab["bmipct"], errors="coerce").isna()
].head(30))

print("\nRader med möjliga extrema värden i bmipct:")
bmipct_num = pd.to_numeric(df_mahtab["bmipct"], errors="coerce")
display(df_mahtab[(bmipct_num > 100) | (bmipct_num < 0)].head(30))
```
