```python
import pandas as pd
import numpy as np
from pathlib import Path

# Hitta projektmappen
project_folder = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()

# Läs in det gemensamma dirty datasetet
df = pd.read_csv(project_folder / "data" / "dirty" / "yrbs_dirty_9vars.csv")

# Välj bara Spirits variabler
spirit_cols = ["row_id", "height_orig", "social_media_use", "age"]
df_spirit = df[spirit_cols].copy()

print("Shape:")
print(df_spirit.shape)

print("\nFörsta 20 raderna:")
display(df_spirit.head(20))

print("\n30 slumpmässiga rader:")
display(df_spirit.sample(30, random_state=42))

print("\nInfo:")
df_spirit.info()

print("\nMissing values:")
print(df_spirit.isnull().sum())

print("\nDubbletter:")
print(df_spirit.duplicated().sum())

print("\nVärden i age:")
print(df_spirit["age"].value_counts(dropna=False))

print("\nVärden i social_media_use:")
print(df_spirit["social_media_use"].value_counts(dropna=False))

print("\nVärden i height_orig:")
print(df_spirit["height_orig"].value_counts(dropna=False).head(30))

print("\nBeskrivning av height_orig:")
print(pd.to_numeric(df_spirit["height_orig"], errors="coerce").describe())

print("\nRader där age inte är numerisk eller saknas:")
display(df_spirit[
    pd.to_numeric(df_spirit["age"], errors="coerce").isna()
].head(30))

print("\nRader där social_media_use inte är numerisk eller saknas:")
display(df_spirit[
    pd.to_numeric(df_spirit["social_media_use"], errors="coerce").isna()
].head(30))

print("\nRader där height_orig inte är numerisk eller saknas:")
display(df_spirit[
    pd.to_numeric(df_spirit["height_orig"], errors="coerce").isna()
].head(30))

print("\nRader med möjliga extrema värden i height_orig:")
height_num = pd.to_numeric(df_spirit["height_orig"], errors="coerce")
display(df_spirit[(height_num > 800) | (height_num < 100)].head(30))
```
