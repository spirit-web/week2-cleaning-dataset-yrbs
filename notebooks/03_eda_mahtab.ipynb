```python
import pandas as pd
from pathlib import Path

project_folder = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()

df_mahtab = pd.read_csv(project_folder / "data" / "clean" / "mahtab_clean.csv")
df_mahtab.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(df_mahtab["sleep_hours"], bins=7)
plt.title("Distribution of Sleep Hours")
plt.show()
```

```python


sns.histplot(df_mahtab["bmipct"], bins=20)
plt.title("Distribution of BMI Percentile")
plt.xlabel("BMI Percentile")
plt.ylabel("Count")
plt.show()
```

```python
sns.boxplot(x="sleep_hours", y="bmipct", data=df_mahtab)
plt.title("BMI vs Sleep Hours")
plt.show()
```

```python
print(df_mahtab.shape)
```

```python
sns.scatterplot(
    x="sleep_hours",
    y="bmipct",
    data=df_mahtab,
    hue="sleep_hours",
    palette="viridis",
    alpha=0.6
)

plt.title("Scatterplot: Sleep Hours vs BMI Percentile")
plt.xlabel("Sleep Hours")
plt.ylabel("BMI Percentile")

plt.legend(title="Sleep Hours")

plt.show()
```

```python

sns.set_style("whitegrid")

plt.figure(figsize=(10, 6))

sns.boxplot(
    x="sleep_hours",
    y="bmipct",
    data=df_mahtab,
    palette="pastel"
)

plt.title("BMI vs Sleep Hours")
plt.show()
```

```python

plt.figure(figsize=(10, 6))

sns.boxplot(
    x="sleep_hours",
    y="bmipct",
    hue="sleep_hours",   # 🔥 fix
    data=df_mahtab,
    palette="viridis",
    legend=False         # 🔥 tar bort onödig legend
)

plt.title("BMI Percentile by Sleep Hours", fontsize=14)
plt.xlabel("Sleep Hours", fontsize=12)
plt.ylabel("BMI Percentile", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

sns.histplot(
    df_mahtab["bmipct"],
    bins=20,
    kde=True,
    color="purple"
)

plt.title("Distribution of BMI Percentile", fontsize=14)
plt.xlabel("BMI Percentile", fontsize=12)
plt.ylabel("Count", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.show()
```

```python
sns.set_style("whitegrid")

plt.figure(figsize=(10, 6))

sns.histplot(
    df_mahtab["bmipct"],
    bins=20,
    kde=True,
    color="#6a5acd"
)

plt.title("BMI Percentile Distribution")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

sns.histplot(
    df_mahtab["sleep_hours"],
    bins=7,
    kde=False,
    color="#6a5acd",
    edgecolor="black"
)

plt.title("Distribution of Sleep Hours", fontsize=14)
plt.xlabel("Sleep Hours", fontsize=12)
plt.ylabel("Count", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

ax = sns.countplot(
    x="sleep_hours",
    hue="sleep_hours",   # 🔥 viktigt
    data=df_mahtab,
    palette="viridis",
    legend=False         # 🔥 tar bort onödig legend
)

for p in ax.patches:
    ax.annotate(
        int(p.get_height()),
        (p.get_x() + p.get_width() / 2, p.get_height()),
        ha='center',
        va='bottom'
    )

plt.title("Distribution of Sleep Hours")
plt.xlabel("Sleep Hours")
plt.ylabel("Count")

plt.show()
```
